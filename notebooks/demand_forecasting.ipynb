# Demand Forecasting & Order Volume Prediction

This notebook demonstrates exploratory data analysis (EDA) for the Demand Forecasting project, showcasing target distribution, seasonal demand trends, and feature importance derived from the LightGBM model.

### 1. Target Distribution & Transformation

Demand data is highly right-skewed. To optimize for the RMSLE metric, we perform a log1p transformation: $\log(1 + \text{OrderVolume})$. This perfectly aligns the model's loss objective with the evaluation metric.

### 2. Daily Demand Trend (2013-2015)

Aggregating total order volume across all fulfillment hubs reveals strong weekly and monthly seasonal components, with major spikes driven by promotional events.

<img src="../plots/demand_trend.png" width="800">

### 3. Model Feature Importance (LightGBM)

The LightGBM ensemble utilizes 126 engineered features. Analysis shows that the expanding mean target encoding (`OOF_Hub_Target`), rolling means, and day-of-week are the most powerful predictors.
